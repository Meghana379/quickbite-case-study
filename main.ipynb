**Missing Value Imputation of Age after outlier treatment using IQR**

```python
Q1=data['age'].quantile(0.25)
Q3=data['age'].quantile(0.75)
IQR=Q3-Q1
Lowerbound_Outlier=Q1-(1.5*IQR)
Upperbound_Outlier=Q3+(1.5*IQR)
Outliers=(data['age']<Lowerbound_Outlier)|(data['age']>Upperbound_Outlier)
data.loc[Outliers,'age']=np.nan
median_age=data['age'].median()
print(f"Calculated Median Age: {median_age}")
df_median_imputed = data.copy()
df_median_imputed['age'] = df_median_imputed['age']. fillna(median_age)

print("\n--- DataFrame After Median Imputation (Age) ---")
print(df_median_imputed)

```

# **TASK 2: Outlier Detection and Removal**

```python
from scipy import stats

cols = ['age', 'delivery_time_min', 'avg_order_value']

# 1. Compare IQR Rule vs. Z-Score Rule (|z| > 3)
print('--- OUTLIER DETECTION SUMMARY ---')
for col in cols:
  series = data[col].dropna()

  # IQR Method
  Q1 = series.quantile(0.25)
  Q3 = series.quantile(0.75)
  IQR = Q3 - Q1
  lower_iqr = Q1 - 1.5 * IQR
  upper_iqr = Q3 + 1.5 * IQR
  iqr_outliers = series[(series < lower_iqr) | (series > upper_iqr)]

  # Z-score Method
  z_scores = np.abs(stats.zscore(series))
  z_outliers = series[z_scores > 3]

  print(f'\nColumn: {col}')
  print(
      f'  IQR bounds: [{lower_iqr:.2f}, {upper_iqr:.2f}] ->'
      f' {len(iqr_outliers)} outliers'
  )
  print(f'  Z-score (|z| > 3): {len(z_outliers)} outliers')

# Store a copy for "Before" visualization
df_before = data[cols].copy()

# 2. Apply Decisions (Handle impossible values & outliers BEFORE imputation)
df_clean = data.copy()

# A. Age: Remove impossible values (e.g., < 18 or > 100) by turning them into NaN
df_clean.loc[(df_clean['age'] < 18) | (df_clean['age'] > 100), 'age'] = np.nan

# B. Delivery Time: Cap (Winsorise) extreme upper values at Q3 + 1.5*IQR (or 99th percentile)
Q1_del = df_clean['delivery_time_min'].quantile(0.25)
Q3_del = df_clean['delivery_time_min'].quantile(0.75)
upper_del = Q3_del + 1.5 * (Q3_del - Q1_del)
df_clean['delivery_time_min'] = df_clean['delivery_time_min'].clip(
    upper=upper_del
)

# C. Avg Order Value: Keep all genuine VIP values unchanged

# 3. Compute Imputation Medians AFTER outlier handling
for col in cols:
  clean_median = df_clean[col].median()
  df_clean[col] = df_clean[col].fillna(clean_median)
  print(f'Clean median used to impute {col}: {clean_median:.2f}')

# 4. Plot Box Plots Before and After
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(12, 10))
fig.suptitle(
    'Box Plots: Before vs. After Outlier Treatment & Imputation', fontsize=14
)

for i, col in enumerate(cols):
  # Before
  axes[i, 0].boxplot(df_before[col].dropna(), vert=False)
  axes[i, 0].set_title(f'{col} (Before)')
  axes[i, 0].grid(True, linestyle='--', alpha=0.5)

  # After
  axes[i, 1].boxplot(df_clean[col], vert=False)
  axes[i, 1].set_title(f'{col} (After)')
  axes[i, 1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

# **TASK 3:Distribution Shapes with KDE Plots**

```python
df_clean['log_avg_order_value'] = np.log(
    df_clean['avg_order_value'].clip(lower=1e-5)
)

# Columns to analyze and plot
plot_cols = [
    'age',
    'customer_rating',
    'delivery_time_min',
    'avg_order_value',
    'log_avg_order_value',
    'annual_spend',
    'orders_last_12m',
]

# 1. Calculate and print skewness
print('--- SKEWNESS SUMMARY (.skew()) ---')
for col in plot_cols:
  skew_val = df_clean[col].skew()
  print(f'{col:<22}: {skew_val:.3f}')

# 2. Plot distributions with Mean, Median, and Mode vertical lines
fig, axes = plt.subplots(nrows=4, ncols=2, figsize=(14, 18))
axes = axes.flatten()

for i, col in enumerate(plot_cols):
  ax = axes[i]
  data = df_clean[col].dropna()

  # Plot histogram + KDE curve
  sns.histplot(data, kde=True, ax=ax, color='skyblue', edgecolor='black')

  # Calculate Mean, Median, and Mode
  mean_val = data.mean()
  median_val = data.median()
  # Round continuous numbers to 1 decimal place to get a meaningful empirical mode
  mode_val = data.round(1).mode()[0]
  skew_val = data.skew()

  # Plot vertical lines
  ax.axvline(
      mean_val, color='red', linestyle='--', linewidth=2, label=f'Mean: {mean_val:.2f}'
  )
  ax.axvline(
      median_val,
      color='green',
      linestyle='-',
      linewidth=2,
      label=f'Median: {median_val:.2f}',
  )
  ax.axvline(
      mode_val,
      color='purple',
      linestyle=':',
      linewidth=2.5,
      label=f'Mode: {mode_val:.2f}',
  )

  ax.set_title(f'{col} (Skewness = {skew_val:.2f})', fontsize=12, fontweight='bold')
  ax.set_xlabel(col)
  ax.set_ylabel('Frequency')
  ax.legend()

# Hide the unused 8th subplot
fig.delaxes(axes[-1])

plt.tight_layout()
plt.show()
```

Mean, median and Mode reacts differently to outliers:


1.  Mean is highly sensitive to higher values which makes it shift towards tail of a right skewed matrix
2.  Median is the middle value of an ordered data set. So there wont be a drastict shift with respect to higher values.
1.  Mode is the most common value of dataset. So it stays where the highest peak of the data is present.

```python
df_encoded = df_clean.copy()

# 1. Ordinal / Label Encoding for 'membership' (Order matters)
membership_order = {'Basic': 0, 'Silver': 1, 'Gold': 2, 'Platinum': 3}
df_encoded['membership_encoded'] = df_encoded['membership'].map(membership_order)

# 2. One-Hot Encoding for 'device_type' and 'acquisition_channel'
# Use drop_first=True to avoid the dummy variable trap (multicollinearity)
ohe_cols = pd.get_dummies(
    df_encoded[['device_type', 'acquisition_channel']],
    drop_first=True,
    dtype=int,
)
df_encoded = pd.concat([df_encoded, ohe_cols], axis=1)

# 3. Target Encoding for 'city' (Mean annual_spend per city)
# Note: In a real ML pipeline, compute this mapping ONLY on X_train to prevent data leakage
city_target_mean = df_encoded.groupby('city')['annual_spend'].mean()
df_encoded['city_target_encoded'] = df_encoded['city'].map(city_target_mean)

# 4. Show original and encoded columns side by side
cols_to_compare = [
    'membership',
    'membership_encoded',
    'device_type',
    'acquisition_channel',
    *ohe_cols.columns.tolist(),
    'city',
    'city_target_encoded',
]

print('--- ORIGINAL VS. ENCODED COLUMNS SIDE BY SIDE ---')
print(df_encoded[cols_to_compare].head(10).to_string())
```



1. Assigning numbers alphabetically (such as Bengaluru = 0, Chennai = 1, Delhi = 2, Mumbai = 3) tricks algorithms into believing there is a mathematical order or distance between cities
2. One-hot encoding 10 cities creates 9 or 10 separate binary columns, while scaling up to 50 or 500 cities adds hundreds of mostly-zero columns (a sparse matrix).
1. If device_type has three categories (iOS, Android, Web), knowing the values of any two columns automatically tells you the third (if iOS = 0 and Android = 0, it must be Web
1. Target encoding uses the outcome variable (annual_spend) to construct an input feature (city_target_encoded), leaking ground-truth label information directly into your predictors

```python
df_sorted = (
    df_clean[['annual_spend']]
    .dropna()
    .sort_values(by='annual_spend', ascending=False)
    .reset_index(drop=True)
)

# 2. Compute running totals and percentages
total_customers = len(df_sorted)
total_revenue = df_sorted['annual_spend'].sum()

df_sorted['customer_pct'] = ((df_sorted.index + 1) / total_customers) * 100
df_sorted['cum_revenue_pct'] = (
    df_sorted['annual_spend'].cumsum() / total_revenue
) * 100

# 3. Find what % of customers bring in 50% and 80% of revenue
cust_for_50 = df_sorted[df_sorted['cum_revenue_pct'] >= 50]['customer_pct'].iloc[0]
cust_for_80 = df_sorted[df_sorted['cum_revenue_pct'] >= 80]['customer_pct'].iloc[0]

# Find how much revenue the top 20% of customers bring in
top_20_count = int(0.20 * total_customers)
rev_from_top_20 = df_sorted['cum_revenue_pct'].iloc[top_20_count - 1]

print(f'Customers contributing 50% of revenue: Top {cust_for_50:.1f}%')
print(f'Customers contributing 80% of revenue: Top {cust_for_80:.1f}%')
print(f'Revenue from Top 20% of customers  : {rev_from_top_20:.1f}%')

# 4. Plot A: Customer-Level Pareto Chart (Bars for Spend + Line for Cumulative %)
fig, ax1 = plt.subplots(figsize=(11, 5))

# Bars for individual customer spend
ax1.bar(
    df_sorted['customer_pct'],
    df_sorted['annual_spend'],
    width=100 / total_customers,
    color='skyblue',
    label='Customer Spend',
)
ax1.set_xlabel('Percentage of Customers (Highest to Lowest Spend)')
ax1.set_ylabel('Annual Spend', color='steelblue')
ax1.set_xlim(0, 100)

# Line for cumulative revenue percentage
ax2 = ax1.twinx()
ax2.plot(
    df_sorted['customer_pct'],
    df_sorted['cum_revenue_pct'],
    color='darkred',
    linewidth=2.5,
    label='Cumulative Revenue %',
)
ax2.axhline(80, color='red', linestyle='--', alpha=0.7)
ax2.axvline(cust_for_80, color='red', linestyle='--', alpha=0.7)
ax2.set_ylabel('Cumulative Revenue (%)', color='darkred')
ax2.set_ylim(0, 105)

plt.title('Pareto Chart: Customer Spend and Cumulative Revenue Share')
plt.tight_layout()
plt.show()

# 5. Plot B: Annual Spend on Normal Scale vs. Log Scale
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Normal Scale
axes[0].hist(df_sorted['annual_spend'], bins=40, color='coral', edgecolor='black')
axes[0].set_title('Annual Spend (Normal Scale)')
axes[0].set_xlabel('Annual Spend')
axes[0].set_ylabel('Number of Customers')

# Log Scale (using log10)
log_spend = np.log10(df_sorted['annual_spend'].clip(lower=1))
axes[1].hist(log_spend, bins=40, color='mediumseagreen', edgecolor='black')
axes[1].set_title('Annual Spend (Log Scale)')
axes[1].set_xlabel('Log10 of Annual Spend')
axes[1].set_ylabel('Number of Customers')

plt.tight_layout()
plt.show()
```

```python
del_data = np.sort(df_clean['delivery_time_min'].dropna().values)
n_del = len(del_data)
del_cdf = np.arange(1, n_del + 1) / n_del

# Calculate exact probabilities and percentiles from the CDF
prob_le_30 = np.mean(del_data <= 30) * 100
prob_gt_45 = np.mean(del_data > 45) * 100
median_del = np.percentile(del_data, 50)
p95_del = np.percentile(del_data, 95)

print('--- DELIVERY TIME CDF ANSWERS ---')
print(f'Probability delivery takes <= 30 min : {prob_le_30:.1f}%')
print(f'Probability delivery takes > 45 min  : {prob_gt_45:.1f}%')
print(f'Median delivery time (50th pct)      : {median_del:.1f} min')
print(f'95th percentile delivery time        : {p95_del:.1f} min')

# Plot PDF (normalized histogram) and Empirical CDF for delivery_time_min
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# PDF Plot
axes[0].hist(
    del_data, bins=35, density=True, color='skyblue', edgecolor='black', alpha=0.8
)
axes[0].set_title('PDF of Delivery Time (Normalized Histogram)')
axes[0].set_xlabel('Delivery Time (minutes)')
axes[0].set_ylabel('Density')
axes[0].grid(True, linestyle='--', alpha=0.4)

# Empirical CDF Plot
axes[1].plot(del_data, del_cdf, color='darkblue', linewidth=2)
axes[1].axvline(30, color='green', linestyle='--', label='30 min')
axes[1].axvline(45, color='red', linestyle='--', label='45 min')
axes[1].axhline(0.50, color='gray', linestyle=':', label='Median (0.50)')
axes[1].axhline(0.95, color='purple', linestyle=':', label='95th Pct (0.95)')
axes[1].set_title('Empirical CDF of Delivery Time')
axes[1].set_xlabel('Delivery Time (minutes)')
axes[1].set_ylabel('Cumulative Share (0 to 1)')
axes[1].legend(loc='lower right')
axes[1].grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()

# --- PART 2: age (Approximately Normal) ---
age_data = np.sort(df_clean['age'].dropna().values)
n_age = len(age_data)
age_cdf = np.arange(1, n_age + 1) / n_age

# Plot PDF and Empirical CDF for age
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# PDF Plot
axes[0].hist(
    age_data, bins=30, density=True, color='lightgreen', edgecolor='black', alpha=0.8
)
axes[0].set_title('PDF of Customer Age (Normalized Histogram)')
axes[0].set_xlabel('Age (years)')
axes[0].set_ylabel('Density')
axes[0].grid(True, linestyle='--', alpha=0.4)

# Empirical CDF Plot
axes[1].plot(age_data, age_cdf, color='darkgreen', linewidth=2)
axes[1].axhline(0.50, color='gray', linestyle=':', label='Median (0.50)')
axes[1].set_title('Empirical CDF of Customer Age')
axes[1].set_xlabel('Age (years)')
axes[1].set_ylabel('Cumulative Share (0 to 1)')
axes[1].legend(loc='lower right')
axes[1].grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```

**PDF**  Shows how crowded or dense values are around a specific point. Where the bars or curve are tall, lots of customers fall into that range.

**CDF** Shows the total share of values at or below a specific point. It starts at 0 (0%) on the far left and climbs steadily up to 1 (100%) on the far right as you include everyone.
